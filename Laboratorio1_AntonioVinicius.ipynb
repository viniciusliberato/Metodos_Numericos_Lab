O método `reshape()` devolve um novo array com os mesmos elementos de `x`, mas com um formato diferente — passamos uma tupla, nesse caso `(2, 3)`, indicando 2 linhas e 3 colunas. O caractere `\n` cria uma nova linha no texto impresso.

O resultado mostra que arrays numpy são especificados como sequência de linhas (row-major), diferente de column-major.Python (e numpy) indexa a partir do 0. Pra pegar o elemento no topo à esquerda de `x_reshape`, usamos `x_reshape[0,0]`.

```python
print('x before we modify x_reshape :\n', x)
print('x_reshape before we modify x_reshape :\n', x_reshape)
x_reshape[0, 0] = 20
print('x_reshape after we modify its top left element :\n', x_reshape)
print('x after we modify top left element of x_reshape :\n', x)
```

Modificar `x_reshape` também modificou `x` porque os dois objetos ocupam o mesmo espaço de memória.A gente acabou de ver que dá pra modificar um elemento de um array. Será que dá pra modificar uma tupla também? Não dá — e tentar gera um erro (uma *exception*).

## Indexação de dados

```python
A = np.array(np.arange(36)).reshape((6, 6))
A
```

`A[2,3]` pega o elemento da segunda linha, terceira coluna (lembrando: Python indexa do 0).

```python
A[2, 3]
```

O primeiro número depois do `[` se refere à linha, o segundo à coluna.### Indexing Rows, Columns, and SubmatricesPra selecionar várias linhas de uma vez, passamos uma lista. `[2,3]` pega a segunda e a quarta linha:

```python
A[[2, 3]]
```

Pra selecionar a segunda e a quarta coluna, passamos `[1,3]` como segundo argumento — e usamos `:` como primeiro argumento pra pegar todas as linhas.

```python
A[:, [1, 3]]
```

Agora, suponha que queremos a submatriz da segunda e quarta linha, junto com a primeira e terceira coluna. Aqui a indexação fica um pouco traiçoeira — é natural tentar assim:

```python
A[[1, 3], [0, 2]]
```

O que aconteceu? A gente recebeu um array unidimensional de tamanho 2, igual a:

```python
np.array([A[1, 0], A[3, 2]])
```

Da mesma forma, o código abaixo também não extrai a submatriz que a gente queria (segunda/quarta linha, primeira/terceira/quarta coluna):

```python
A[[1, 3], [0, 2, 3]]
```

O que houve: quando passamos duas listas de índice, o numpy entende que são pares (i,j) de posições individuais — por isso as duas listas precisam ter o mesmo tamanho. Só que não era isso que a gente queria, já que buscávamos uma submatriz.Um jeito simples de fazer certo: primeiro criar uma submatriz filtrando as linhas de `A`, e depois, em cima dessa, filtrar as colunas.

```python
A[[1, 3]][:, [0, 2]]
```

Existem jeitos mais eficientes de chegar no mesmo resultado. A função `np.ix_()` permite extrair uma submatriz usando listas, criando um objeto de mesh intermediário.

```python
idx = np.ix_([2, 1], [2, 4, 5])
A[idx]
```

Também dá pra selecionar submatrizes de forma eficiente usando *slices*. `1:2:1` pega o segundo e o quarto item de uma sequência, e `0:3:2` pega o primeiro e o terceiro (o terceiro número do slice é o passo).

```python
A[1:2:1, 0:3:2]
```

Por que dá pra pegar a submatriz direto com *slices*, mas não com listas? São tipos diferentes de Python, tratados de forma diferente pelo numpy. Slices funcionam em qualquer sequência (strings, listas, tuplas); listas como índice são mais limitadas.### Boolean IndexingUm booleano é um tipo que vale `True` ou `False` (também representado como 1 e 0). A linha abaixo cria um vetor de `False`, do tamanho da primeira dimensão de `A`.

```python
keep_rows = np.zeros(A.shape[1], bool)
keep_rows
```

Agora colocamos dois desses elementos como `True`:

```python
keep_rows[[0, 2]] = True
keep_rows
```

Os elementos de `keep_rows`, vistos como inteiros, são iguais a `np.array([0,1,0,1])`. Usamos `==` pra conferir isso — aplicado a dois arrays, `==` roda elemento por elemento.

```python
np.all(np.isin(keep_rows, [2, 3, 1, 4]))
```

(`np.all()` conferiu se todos os elementos de um array são `True`. Existe também `np.any()`, que confere se algum elemento é `True`.)Só que, mesmo `np.array([0,1,0,1])` e `keep_rows` sendo iguais pelo `==`, eles indexam linhas diferentes! O primeiro pega a primeira, segunda, primeira e segunda linha de `A`.

```python
A[np.array([0, 1, 0, 1])]
```

Já `keep_rows` pega só a segunda e a quarta linha — as linhas onde o booleano é `True`.

```python
A[keep_rows]
```

Isso mostra que booleanos e inteiros são tratados de forma diferente pelo numpy.De novo usando `np.ix_()`, agora criamos um mesh com a segunda e quarta linha e a primeira, terceira e quarta coluna — dessa vez aplicando a função a booleanos, não listas.

```python
keep_cols = np.zeros(A.shape[1], bool)
keep_cols[[0, 2, 3]] = True
idx_bool = np.ix_(keep_rows, keep_cols)
A[idx_bool]
```

Também dá pra misturar uma lista com um array de booleanos nos argumentos de `np.ix_()`:

```python
idx_mixed = np.ix_([3, 4], keep_cols)
A[idx_mixed]
```

## Loop FOR

```python
total = 0
for value in [4, 10, 20]:
    total += value
print('Total is: {0}'.format(total))
```

O código indentado abaixo do `for` roda pra cada valor da sequência. O loop termina quando a célula acaba ou quando o código volta a ficar no mesmo nível de indentação do `for`. Dá pra aninhar loops com indentação extra.

```python
total = 0
for value in [4, 10, 20]:
    for weight in [5, 20, 10]:
        total += value * weight
print('Total is: {0}'.format(total))
```

Aqui somamos cada combinação de `value` e `weight`. Também usamos a notação de incremento: `a += b` equivale a `a = a + b`.Uma tarefa mais comum é somar pares (value, weight) — por exemplo, calcular a média de uma variável que assume valores 2, 3 ou 19 com probabilidade 0.2, 0.3, 0.5. Pra isso dá pra usar `zip()`, que percorre uma sequência de tuplas.

```python
total = 0
for value, weight in zip([4, 10, 20], [5, 20, 10]):
    total += weight * value
print('Weighted average is: {0}'.format(total))
```
